```python
import pandas as pd

data = {
    "Customer_ID": [101, 102, 103, 104, 105,
                    106, 107, 108, 109, 110],
    "Name": ["Aman", "Priya", "Rahul", "Neha", "Karan",
             "Simran", "Arjun", "Riya", "Vikas", "Anjali"],
    "Age": [25, 32, 28, 35, 22, 40, 31, 27, 45, 29],
    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],
    "Purchase_Amount": [2000, 3500, 1500, 4000, 2500,
                        3000, None, 4500, 1800, 3200]
}

df = pd.DataFrame(data)
print("First 5 rows:")
print(df.head())
print(df.describe())
print("\nMissing Values:")
print(df.isnull().sum())
df["Purchase_Amount"] = df["Purchase_Amount"].fillna(
    df["Purchase_Amount"].median()
)
print("\nData after handling missing values:")
print(df)
print("\nMissing Values After Handling:")
print(df.isnull().sum())
```

```python
older_customers = df[df["Age"] > 30]

print(older_customers)
```

```python
gender_total = df.groupby("Gender")["Purchase_Amount"].sum()

print(gender_total)
```

```python
selected_data = df[
    ["Customer_ID", "Name", "Purchase_Amount"]
]

print(selected_data)
```

```python
df["Discounted_Price"] = df["Purchase_Amount"] * 0.90

print(df)
```

```python
gender_average = df.groupby("Gender")["Purchase_Amount"].mean()

print(gender_average)
```

```python
sorted_df = df.sort_values(
    by="Purchase_Amount",
    ascending=False
)

print(sorted_df)
```

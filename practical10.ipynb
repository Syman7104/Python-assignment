```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_iris

print("All libraries imported successfully!")
```

```python
# Load Iris dataset
data = load_iris()

# Convert dataset into DataFrame
iris = pd.DataFrame(
    data.data,
    columns=[
        'sepal_length',
        'sepal_width',
        'petal_length',
        'petal_width'
    ]
)

# Add flower species
iris['species'] = pd.Categorical.from_codes(
    data.target,
    data.target_names
)

# Display first five rows
print(iris.head())
```

```python

# Scatter Plot

sns.scatterplot(
    data=iris,
    x='sepal_length',
    y='sepal_width',
    hue='species'
)

plt.title('Scatter Plot of Iris Dataset')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')

plt.show()
```

```python
# Density Plot

sns.kdeplot(
    data=iris,
    x='sepal_length',
    hue='species',
    fill=True
)

plt.title('Density Plot of Sepal Length')
plt.xlabel('Sepal Length')
plt.ylabel('Density')

plt.show()
```

```python
# Violin Plot

sns.violinplot(
    data=iris,
    x='species',
    y='petal_length'
)

plt.title('Violin Plot of Petal Length')
plt.xlabel('Flower Species')
plt.ylabel('Petal Length')

plt.show()
```

```python
# Heatmap

correlation = iris.select_dtypes(
    include='number'
).corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Iris Dataset')

plt.show()
```
